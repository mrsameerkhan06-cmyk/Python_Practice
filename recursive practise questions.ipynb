```python
#wap to even or odd in function
def even_odd(n):
  if(n%2==0):
    print(n,"is even")
  else:
    print(n,"is odd")

even_odd(11)
```

```python
n = int(input("enter the element "))
def f(n):
  if(n==0):
    return
  else:
    f(n-1)
    print(n)

f(n)
```

```python
#factoral of n number funtion
def fact(n):
  if (n == 0 or n==1):
    return 1
  else:
    return n+fact(n-1)


fact(4)
```

```python
# sum of first n numbers
sum = 0
def Sum(n):
  if(n == 0):
    return 0
  else:
    sum = n+Sum(n-1)
    return sum

d= int(input("enter number : "))
Sum(d)
```

```python
#write a recursive function print a;; element in list #hint :- use list & index as parameters
def printidx( list, idx = 0):
  print(list[0])
  printidx(list[1:],idx+1)
  return idx

w = [1,2,3,4,5]
printidx(w,)
```
