```python
marks = float(input("enter your marks :"))

if(marks >=95 and marks <=100):
  print("grade 'A++'")
elif(marks <=95 and marks >=90):
  print("grade 'A+'")
elif(marks <90 and marks >=83):
  print("grade 'A'")
elif(marks <83 and marks >=70):
  print("grade 'B+'")
elif(marks <70 and marks >=60):
  print("grade 'B'")
elif(marks <60 and marks >=50):
  print("grade 'C+'")
elif(marks <50 and marks>40):
  print("Grade 'C'")
elif(marks <40 and marks>=34):
  print("Grade 'D'")
else:
  print("FAIL... Try again next SEM")
```
