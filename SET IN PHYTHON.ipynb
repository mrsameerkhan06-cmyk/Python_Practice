```python
#SET In Phython
#
#METHODS IN PHYHTON
set1= {1, 2}
set1.add(1)
set1.add(2)

set1.clear()  #clear the set values comletly
set1.add("sameer")  #adds valuess in set randomnly
set1.add("khan")

set1.pop() #randomnly pop/removes values
SET2 = (1, 3, 4, "SAMEER")

print(set1.union(SET2)) #COMBINES COMMON VALUES & RETURN

print(set1.intersection(SET2))
print(type(set1))
print(set1)

```
