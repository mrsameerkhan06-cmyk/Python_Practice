```python
#print the perfect square number upto 10 in py
i = 1
while i<=10:
  print(i*i)
  i+=1
```

```python
##print list elements which are perfect square in list using loop
#[1, 4, 9, 16, 25, 36, 49, 64, 81, 100]
li=[1, 4, 9, 16, 25, 36, 49, 64, 81, 100]
i = li[0]
print(i) # to test the indexing
while i<=len(li)-1:
  print(li[i])
  i+=1
```
